### 1. Employee Salary Adjustment

```python
salaries = np.array([45000, 52000, 61000, 48000, 75000])
salaries
```

##### a. Increase all salaries by 10%.

```python
incremented_salaries = salaries + salaries / 10
incremented_salaries
```

##### b. Find the average salary.

```python
print(f'Average salary before increment: {salaries.mean()}')
print(f'Average salary after increment: {incremented_salaries.mean()}')
```

##### c. Find the maximum and minimum salary.

```python
print('Maximum and Minimum Salaries Before Increment')
print(f'Minimum Salary: {salaries.min()}')
print(f'Maximum Salary: {salaries.max()}')

print()

print('Maximum and Minimum Salaries After Increment')
print(f'Minimum Salary: {incremented_salaries.min()}')
print(f'Maximum Salary: {incremented_salaries.max()}')
```

##### d. Print salaries greater than 60000

```python
salaries_greater_than_60k = salaries[salaries > 60000]
print(f'Salaries greater than 60K before increment: {salaries_greater_than_60k}')

print()

salaries_greater_than_60k = incremented_salaries[incremented_salaries > 60000]
print(f'Salaries greater than 60K after increment: {salaries_greater_than_60k}')
```

### 2. Sales Data Analysis

Daily sales of a store for 10 days:

```python
sales = np.array([120,150,180,200,170,160,210,220,190,175])
sales
```

##### a. Compute total sales

```python
total_sales = sales.sum()
print(f'Total Sales: {total_sales}')
```

##### b. Compute mean and standard deviation

```python
sales_mean = sales.mean()
sales_sd = sales.std()

print(f'Sales Mean = {sales_mean}')
print(f'Sales Standard Deviation = {sales_sd}')
```

##### c. Identify days where sales exceeded the average.

```python
days_with_sales_more_than_avg = np.where(sales > sales_mean)[0]
print(f'Days when sales exceeded the average: ')
for day in days_with_sales_more_than_avg:
  print(f'Day {day + 1}', end = ', ')
```

##### d. Normalize the sales using Min-Max scaling.

```python
sales_min = sales.min()
sales_max = sales.max()

sales_normalized = (sales - sales_min) / (sales_max - sales_min)

print(sales_normalized)
```

### 3. Matrix Operations for Business Data

A company has sales data for 3 products across 4 regions.

```python
sales_matrix = np.array([
[120,140,130,150],
[80,90,100,95],
[200,210,190,205]
])
```

##### a. Find total sales per product.

```python
total_sales_per_product = sales_matrix.sum(axis=1)
total_sales_per_product
for i in range(total_sales_per_product.size):
  print(f'Total sales for Product {i + 1} = {total_sales_per_product[i]}')
```

##### b. Find total sales per region.

```python
total_sales_per_region = sales_matrix.sum(axis=0)
total_sales_per_region
for i in range(total_sales_per_region.size):
  print(f'Total sales for Region {i + 1} = {total_sales_per_region[i]}')
```

##### c. Identify the best performing product.

```python
print('Best performing products: ')
best_performing_products = np.where(total_sales_per_product == total_sales_per_product.max())[0]
for product_index in best_performing_products:
  print(f'Product {product_index + 1}')
```

##### d. Identify the region with highest sales.

```python
print('Regions with highest sale: ')
highest_sale_regions = np.where(total_sales_per_region == total_sales_per_region.max())[0]
for region_index in highest_sale_regions:
  print(f'Region {region_index + 1}')
```

### 7. Customer Purchase Analysis

```python
cust_data = {
'CustomerID':[101,102,103,104,105,106,107],
'Age':[25,45,35,50,23,40,60],
'PurchaseAmount':[250,600,400,800,150,700,900],
'City':['Delhi','Mumbai','Delhi','Chennai','Mumbai','Delhi','Chennai']
}
```

Converting to data frame:

```python
cust_df = pd.DataFrame(cust_data)
cust_df
```

##### a. Find the average purchase amount by city.

```python
cust_df.groupby('City')['PurchaseAmount'].mean().rename('Average Purchase Amount')
```

##### b. Identify customers with purchase above overall average.

```python
#mean purchase amount
mean_purchase_amount = cust_df['PurchaseAmount'].mean()
print(f'Mean Purchase Amount = {mean_purchase_amount}')
```

```python
cust_df[cust_df['PurchaseAmount'] > mean_purchase_amount]
```

##### c. Find the city generating the highest revenue.

```python
city_revenues = cust_df.groupby('City')['PurchaseAmount'].sum()
city_revenues
```

Highest Revenue generating Cities:

```python
city_revenues[city_revenues == city_revenues.max()]
```

##### d. Create a column Age_Group:

*   <30 → Young
*   30 – 50 → Adult
*   \>50 → Senior






```python
age_bins = [0, 30, 50, float('inf')]
age_labels = ['Young', 'Adult', 'Senior']
cust_df['Age_Group'] = pd.cut(cust_data['Age'], bins = age_bins, labels = age_labels, right = False)
cust_df
```

##### e. Find average purchase per age group.

```python
cust_df.groupby('Age_Group', observed=True)['PurchaseAmount'].mean().rename('Average Purchase Amount')
```

### 8. Time Series Sales Analysis

Generate data:

```python
dates = pd.date_range('2024-01-01', periods=12, freq='M')
sales = [200,220,250,270,300,320,350,360,380,400,420,450]
df = pd.DataFrame({'Date':dates,'Sales':sales})
```

##### a. Convert Date to datetime index.

```python
df.set_index('Date', inplace = True)
df
```

##### b. Compute monthly percentage growth.

```python
df['MonthlyPercentageGrowth'] = df['Sales'].pct_change() * 100
df
```

##### c. Calculate 3-month rolling average sales.

```python
df['3M_Rolling_Avg'] = df['Sales'].rolling(window=3).mean()
df
```

##### d. Identify the month with highest growth rate.

```python
df[df['MonthlyPercentageGrowth'] == df['MonthlyPercentageGrowth'].max()]
```
