# 1. Bank Account Manager

#### a. Simulate a bank account with deposit, withdrawal, and balance enquiry operations

```python
import datetime

# creating a class to hold bank account details

class BankAccount:

  def __init__(self, first_name, last_name, account_no, branch_name, balance, dob):
    self.first_name = first_name
    self.last_name = last_name
    self.account_no = account_no
    self.branch_name = branch_name
    self.balance = balance
    self.dob = dob
    self.full_name = first_name + ' ' + last_name

  def __str__(self):
    return f'Account Holder: {self.full_name} \nAccount No: {self.account_no} \nBranch Name: {self.branch_name} \nDOB: {self.dob} \nAccount Balance: ₹{self.balance}'

  def deposit(self, amount):
    self.balance += amount
    print(f'Deposited ₹ {amount}')
    print(f'Updated balance: ₹{self.balance}')

  def check_balance(self):
    print(f'Account holder: {self.full_name}')
    print(f'Account No: {self.account_no}')
    print(f'Current balance: ₹{self.balance}')

  def withdraw(self, amount):
    if (amount > self.balance):
      print('Insufficient balance')
      return

    self.balance -= amount
    print(f'{amount} withdrawed succesfully!')
    print(f'Current balance: ₹{self.balance}')


```

#### c. Implement conditional statements to determine weather conditions (sunny, rainy, etc.)

```python
"""
Returns weather condition based on temperature, humidity and wind values
"""
def get_weather_condition(temp, humidity, wind):

    if wind >= 15:
        return "Windy"

    elif humidity >= 80 and wind >= 8:
        return "Rainy"

    elif temp >= 32 and humidity < 70:
        return "Sunny"

    elif humidity >= 75:
        return "Cloudy"

    else:
        return "Partly Cloudy"


print('Weather conditions in the month of February in Kannur:\n')
for date, data in knr_weather_data.items():
  print(f'{date}: {get_weather_condition(data['temperature_in_celsius'], data['humidity'], data['wind'])}')
```

#### d. Use functions to calculate average temperature and humidity, and print results

```python
# function to calculate average temperature

def show_avg_temperature(temperature_values):
  avg_temp = sum(temperature_values) / len(temperature_values)
  print(f'Average Temperature: {avg_temp} °C')

# function to calculate average humidity
def show_avg_humidity(humidity_values):
  avg_humidity = sum(humidity_values) / len(humidity_values)
  print(f'Average humidity: {avg_humidity} %')

# function to calculate average wind
def show_avg_wind(wind_values):
  avg_wind = sum(wind_values) / len(wind_values)
  print(f'Average wind: {avg_wind} km/h')
```

##### Extracting temperature, humidity and wind values

```python
temp_values = [data["temperature_in_celsius"] for data in knr_weather_data.values()]
humidity_values =  [data["humidity"] for data in knr_weather_data.values()]
wind_values =  [data["wind"] for data in knr_weather_data.values()]
```
