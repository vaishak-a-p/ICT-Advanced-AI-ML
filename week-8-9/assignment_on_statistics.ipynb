### 10. Z-Test: Assume the national average math score is 50. Use a one-sample z-test to test if this sample's average math score is significantly different from the national average.

```python
# One-Sample Z-Test: Math Score vs National Average

# We test whether the sample mean math score
# is significantly different from the national average (50)

import numpy as np
from scipy.stats import norm
```

```python
# Step 1: Define population mean (national average)
mu0 = 50

# Sample mean (x̄)
x_bar = df['math score'].mean()

# Sample standard deviation (s)
s = df['math score'].std()

# Sample size (n)
n = df['math score'].count()

print("Sample mean:", x_bar)
print("Sample std:", s)
print("Sample size:", n)
```

```python
# Step 3: Compute Z-score

# Formula:
# z = (x̄ - μ0) / (s / sqrt(n))

z = (x_bar - mu0) / (s / np.sqrt(n))

print("Z-score:", z)
```

```python
# Step 4: Compute p-value (two-tailed test)

# We multiply by 2 because it's a two-sided test
p_value = 2 * (1 - norm.cdf(abs(z)))

print("P-value:", p_value)
```

```python
# Step 5: Decision

alpha = 0.05  # significance level

if p_value < alpha:
    print("Reject H0: Mean is significantly different from 50")
else:
    print("Fail to reject H0: No significant difference from 50")
```

### 11. T-Test: Compare the reading scores between students with different lunch types (free/reduced vs. standard) to test if there is a significant difference.

```python
# Independent Two-Sample T-Test

import numpy as np
from scipy.stats import ttest_ind
```

```python
# Step 1: Split the data into two groups

# Students with free/reduced lunch
group_free = df[df['lunch'] == 'free/reduced']['reading score']

# Students with standard lunch
group_standard = df[df['lunch'] == 'standard']['reading score']

print("Free/Reduced count:", len(group_free))
print("Standard count:", len(group_standard))
```

```python
# Step 2: Compute basic statistics (optional but useful)

print("Free/Reduced mean:", group_free.mean())
print("Standard mean:", group_standard.mean())
```

```python
# Step 3: Perform Independent T-Test

# H0: No difference in reading scores between groups
# H1: There is a difference

t_stat, p_value = ttest_ind(group_free, group_standard)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
# Step 4: Decision Rule

alpha = 0.05  # significance level

if p_value < alpha:
    print("Reject H0: Significant difference in reading scores")
else:
    print("Fail to reject H0: No significant difference")
```

### 12. F-Test: Check if the variances of math scores differ significantly between students with and without test preparation.

```python
# F-Test for Variance Comparison

import numpy as np
from scipy.stats import f
```

```python
# Step 1: Split the data into two groups

# Students who completed test preparation
prep_yes = df[df['test preparation course'] == 'completed']['math score']

# Students who did NOT complete test preparation
prep_no = df[df['test preparation course'] == 'none']['math score']

print("Completed prep count:", len(prep_yes))
print("No prep count:", len(prep_no))
```

```python
# Step 2: Compute sample variances

var_yes = np.var(prep_yes, ddof=1)  # sample variance
var_no = np.var(prep_no, ddof=1)

print("Variance (completed):", var_yes)
print("Variance (none):", var_no)
```

```python
# Step 3: Compute F-statistic

# Always put larger variance in numerator
F = var_yes / var_no if var_yes > var_no else var_no / var_yes

print("F-statistic:", F)
```

```python
# Step 4: Degrees of freedom

df1 = len(prep_yes) - 1
df2 = len(prep_no) - 1

print("df1:", df1)
print("df2:", df2)
```

```python
# Step 5: Compute p-value

# We use survival function (1 - cdf)
p_value = 1 - f.cdf(F, df1, df2)

# Two-tailed adjustment
p_value *= 2

print("P-value:", p_value)
```

```python
# Step 6: Decision Rule

alpha = 0.05  # significance level

if p_value < alpha:
    print("Reject H0: Variances are significantly different")
else:
    print("Fail to reject H0: No significant difference in variances")
```
